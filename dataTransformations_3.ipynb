# Handling Nulls

## Dropping Nulls

dropna(how: any) -> any means, if there is a null value in any column, it will delete the record instantly without checking any other things

dropna(how: all) -> all means, if there is null value in all of the column, then only it will delete the record

```python
df.dropna('all').limit(10).toPandas()
```

```python
df.dropna('any').toPandas()
```

almost half of the records are deleted because originally we had around 8K rows and now after dropping, 4K results are shown.

```python
df.dropna(subset= ['Outlet_Size']).toPandas()
```

### this function is preferred because we can input which column to look for before deleting the records, we can give the list of columns to check for null and delete in subset

## Filling the Null Values

```python
df.toPandas()
```

### Scenario 1: Filling all the NaN (Null) cells with 'Not Available' value

```python
df.fillna('Not Available').toPandas()
```

## As in Dropping Nulls, we can pass the subset columns where we want to make the fill effective leaving other columns as it as

### Example: We are filling the NaN of 'Outlet_Size' column only to 'Not Available'

```python
df.fillna('Not Available', subset= ['Outlet_Size']).toPandas()
```

# Split and Indexing

### If we want to separate the values of a single column for some reason, for e.g. in this data fram, we have 'Supermarket Type 1', Type 2 and so on, we can make use of Split function to separat them and store the values as a list. In order to fetch them later, we can make use of Indexing

```python
df.withColumn('Outlet_Type', F.split('Outlet_Type', ' ')).toPandas()
```

>> Column 'Outlet_Type' is converted into the list of its original values separated by a space 

### Now with the help of Index, we are going to fetch only either the Outlet Type is Type 1 or Type 2 or Store... like that

```python
df.withColumn('Outlet_Type', F.split('Outlet_Type', ' ')[1]).limit(10).toPandas()
```

# Array Contains

### If we want to find out if any particulr value is available in the splitted column, we can make use of Array Contains

```python
df_explode.limit(10).toPandas()
```

```python
df_explode.withColumn('Type1 Availability', F.array_contains('Outlet_Type', 'Type1')).limit(10).toPandas()
```

### >> Type 1 Availability column gives True whenever the Outlet_Type column records list has 'Type1'

# Group By

### Calculate the total MRP Price for each item type

```python
type('Item_MRP')
```

```python
df=df.withColumn('Item_MRP', F.col('Item_MRP').cast(T.DoubleType()))
```

```python
df.dtypes
```

```python
df.groupBy('Item_Type').agg(F.sum('Item_MRP')).toPandas()
```

### Likewise, let's find out the average MRP of each item type

```python
df.groupBy('Item_Type').agg(F.avg('Item_MRP').alias('Average Item MRP')).toPandas()
```

### Let's try grouping two columns at once

```python
df.groupBy('Item_Type', 'Outlet_Size').agg(F.sum('Item_MRP').alias('Total MRP')).toPandas()
```

### Let's level up by adding one more column in the aggregation i.e. AVG

```python
df.groupBy('Item_Type', 'Outlet_Size').agg(F.sum('Item_MRP'), F.avg('Item_MRP')).toPandas()
```

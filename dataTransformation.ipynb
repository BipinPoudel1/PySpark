```python
from pyspark.sql import SparkSession
import pyspark.sql.types as T
```

```python
spark= SparkSession.builder\
        .appName("Data Transformation in PySpark")\
        .master("local[*]")\
        .getOrCreate()
```

```python
df= spark.read.format('csv').option('inferSchema', True).option('header', True)\
    .load('pathToData')
```

```python
df.toPandas()
```

### SELECT Method 1

```python
df_sel= df.select('Item_Identifier', 'Item_Weight', 'Item_Fat_Content')
```

```python
df_sel.toPandas()
```

### Method 2 using COL

```python
from pyspark.sql.functions import *
```

```python
df.select(col('Item_Identifier'), col('Item_Weight'), col('Item_Fat_Content')).toPandas()
```

## ALIAS

```python
df.select(col('Item_Identifier').alias('Item ID')).toPandas()
```

## FILTER/ WHERE

# We are going to:

### 1. Filter the data with fat content = regular
### 2. Slice the data with item type= Soft Drinks and Weight < 10
### 3. Fetch the data with Tier in (Tier 1 or Tier 2) and Outlet Size is Null

### Scenario 1 here

```python
df.filter(col('Item_Fat_Content')=='Regular').toPandas()
```

### Scenario 2: Slice the data with item type= Soft Drinks and Weight < 10

```python
df.filter(col('Item_Type')=='Soft Drinks').filter(col('Item_Weight')< 10).toPandas()
```

### or even better one using AND Operator (&) in PySpark

```python
df.filter((col('Item_Type')=='Soft Drinks') & (col('Item_Weight')<10)).toPandas()
```

### Scenario 3: Fetch the data with Tier in (either Tier 1 or Tier 2) and Outlet Size is Null

```python
df.filter(col('Outlet_Size').isNull() & (col('Outlet_Location_Type').isin('Tier 1', 'Tier 2'))).toPandas()
```

## withColumnRenamed

### changing the name of the column 'Item Weight' to 'Item Wt.'

```python
df.withColumnRenamed('Item_weight', 'Item Wt.').toPandas()
```

## withColumn

```python
df= df.withColumn('flag', lit("new"))
```

```python
df.toPandas()
```

df.withColumn('Multiply', col('Item_Weight')*col('Item_MRP')).toPandas()

### Change the content of Item_Fat_Content onto their short forms like LF, R, ...

```python
df.withColumn('Item_Fat_Content', regexp_replace(col('Item_Fat_Content'), "Regular", "R"))\
    .withColumn('Item_Fat_Content', regexp_replace(col('Item_Fat_Content'), "Low Fat", "LF")).toPandas()
```

# Type Casting

```python
df= df.withColumn('Item_Weight', col('Item_Weight').cast(T.StringType()))
```

```python
df.toPandas()
```

```python
df.printSchema()
```

# Sort/ Order By

### Sorting our data from highest to lowest Item weight

```python
df.sort(col('Item_weight').desc()).toPandas()
```

```python
df.sort(col('Item_Visibility').asc()).toPandas()
```

### Sorting based on multiple columns

```python
df.sort(['Item_Weight', 'Item_Visibility'], ascending=[0, 0]).toPandas()
```

### Sorting by ascending order in one column and by descending order in another column

```python
df.sort(['Item_Weight', 'Item_Visibility'], ascending=[0, 1]).toPandas()
```

```python
df.limit(10).toPandas()
```
